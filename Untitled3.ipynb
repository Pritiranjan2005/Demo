```python
import pyspark
import pandas as pd
```

```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName('Practice1').getOrCreate()
```

```python
df_pyspark=spark.read.csv('/content/class3.csv',inferSchema=True,header=True,nullValue='null')
```

```python
df_pyspark.show()
```

```python
df_pyspark.drop('Name').show()
```

```python
#drop null rows
df_pyspark.na.drop().show()
```

```python
##any==how
df_pyspark.na.drop(how='all').show()
```

```python
##threshold
df_pyspark.na.drop(how='any',thresh=2).show()
```

```python
##subset
df_pyspark.na.drop(how='any',subset=['Experience']).show()
```

```python
###filling the missing value
df_pyspark.na.fill('Missing Values').show()
```

```python
# if you fill the missing value of a specific column
df_pyspark.na.fill('Missing Values','Experience').show()
```

```python
df_pyspark.show()
```

```python
## fill the experience with mean of the experience
from pyspark.ml.feature import Imputer
imputer = Imputer(
    inputCols=['Age','Experience','Salary'],
    outputCols=["{}_imputed".format(c) for c in ['age','Experience','Salary']]
).setStrategy("mean")
```

```python
## add imputation cols to df
imputer.fit(df_pyspark).transform(df_pyspark).show()
```

```python
##class-4
#filter operation &,|,==,~
df_pyspark=spark.read.csv('/content/filter.csv',header=True,inferSchema=True)
df_pyspark.show()



```

```python
## salary of the people less than or equal to 20000
df_pyspark.filter("Salary<=20000").show()
```

```python
df_pyspark.filter("Salary<=20000").select(['Name','age']).show()
```

```python
df_pyspark.filter(df_pyspark['Salary']<=20000).show()
```

```python
df_pyspark.filter((df_pyspark['Salary']<=20000) & (df_pyspark['Salary']>=15000)).show()
```

```python
df_pyspark.filter((df_pyspark['Salary']<=20000) | (df_pyspark['Salary']>=15000)).show()
```

```python
df_pyspark.filter(~(df_pyspark['Salary']<=20000)).show()
```
